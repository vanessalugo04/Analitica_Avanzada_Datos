```python
df = pd.read_csv('denue_09_csv\conjunto_de_datos\denue_inegi_09_.csv', sep=',', encoding='latin1')

df.head()
df.info()
```

```python
# Estandarizar texto
df['nombre_act'] = df['nombre_act'].str.upper().str.strip()
df['nom_vial'] = df['nom_vial'].str.upper().str.strip()

# Eliminar nulos importantes
df = df.dropna(subset=['latitud', 'longitud'])

# Duplicados
df = df.drop_duplicates()
df.shape
```

```python
coords = df[['latitud', 'longitud']].values
nn = NearestNeighbors(radius=0.01)
nn.fit(coords)

distancias, indices = nn.radius_neighbors(coords)

df['competencia'] = [len(i) for i in indices]
```

```python
df['competencia'].hist()
plt.title("Distribución de competencia")
plt.show()
```

```python
df['lat_bin'] = pd.cut(df['latitud'], bins=20)
df['lon_bin'] = pd.cut(df['longitud'], bins=20)

df['zona'] = df['lat_bin'].astype(str) + "_" + df['lon_bin'].astype(str)
```

```python
zonas = df.groupby('zona').agg({
    'id': 'count',
    'competencia': 'mean'
}).rename(columns={'id': 'total_negocios'})
```

```python
dominante = df.groupby('zona')['nombre_act'] \
    .agg(lambda x: x.value_counts().idxmax())

zonas['actividad_dominante'] = dominante
```

```python
df.groupby('nombre_act')['competencia'].mean().sort_values()
```

```python
df.groupby('per_ocu')['competencia'].mean()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, roc_auc_score
```

```python
zonas['score'] = (
    (zonas['total_negocios'] / zonas['total_negocios'].max() * 0.40) +
    (zonas['actividad_dominante'].map(lambda x: 1) * 0.30) +
    (1 - zonas['competencia'] / zonas['competencia'].max() * 0.30)
)
 
umbral = zonas['score'].quantile(0.75)
zonas['es_atractiva'] = (zonas['score'] > umbral).astype(int)
 
print(f"\nVariable objetivo creada:")
print(f"  Zonas atractivas: {zonas['es_atractiva'].sum()}")
print(f"  Zonas no atractivas: {(1 - zonas['es_atractiva']).sum()}")
```

```python
X = zonas[['total_negocios', 'competencia']].copy()
y = zonas['es_atractiva'].copy()
X = X.fillna(X.mean())
 
print(f"\nFeatures: {list(X.columns)}")
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
 
modelo = LogisticRegression(random_state=42, max_iter=1000)
modelo.fit(X_scaled, y)
```

```python
 
y_pred = modelo.predict(X_scaled)
y_proba = modelo.predict_proba(X_scaled)[:, 1]
 
accuracy = accuracy_score(y, y_pred)
auc = roc_auc_score(y, y_proba)
cm = confusion_matrix(y, y_pred)
 

print(f"RESULTADOS")
print(f"Accuracy: {accuracy:.2%}")
print(f"AUC-ROC: {auc:.4f}")
 
print(f"\nMatriz de confusión:")
print(f"           Pred NO  Pred SÍ")
print(f"Real NO      {cm[0,0]:2d}      {cm[0,1]:2d}")
print(f"Real SÍ      {cm[1,0]:2d}      {cm[1,1]:2d}")
```

```python
print(f"\nCoeficientes:")
for feature, coef in zip(X.columns, modelo.coef_[0]):
    signo = "↑" if coef > 0 else "↓"
    print(f"  {signo} {feature}: {coef:+.4f}")
```

```python
zonas['prob_atractiva'] = y_proba
zonas['prediccion'] = y_pred
 
print(f"TOP 5 ZONAS")
 
top_5 = zonas.nlargest(5, 'prob_atractiva')
for idx, (zona_id, row) in enumerate(top_5.iterrows(), 1):
    print(f"\n{idx}. Zona {zona_id}")
    print(f"   Negocios: {int(row['total_negocios'])}")
    print(f"   Competencia: {row['competencia']:.2f}")
    print(f"   Probabilidad: {row['prob_atractiva']*100:.1f}%")
```
