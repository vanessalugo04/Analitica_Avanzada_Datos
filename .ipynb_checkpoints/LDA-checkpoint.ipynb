# Librerías

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from numpy.linalg import inv
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

# Crear conjunto de datos

```python
np.random.seed(7)
alumnos = 30

x1_aprob = np.random.normal(loc = 15, scale = 2, size = alumnos)
x2_aprob = np.random.normal(loc = 90, scale = 5, size = alumnos)

x1_reprob = np.random.normal(loc = 5, scale = 2, size = alumnos)
x2_reprob = np.random.normal(loc = 60, scale = 5, size = alumnos)
```

# Conjunto de datos de un dataframe

```python
df = pd.DataFrame({
    'Alumno': [f'A{i+1}' for i in range(alumnos)] + [f'R{i+1}' for i in range(alumnos)],
    'Horas_estudio': np.concatenate([x1_aprob, x1_reprob]),
    'Asistencia': np.concatenate([x2_aprob, x2_reprob]),
    'Clase': ['Aprobado'] * alumnos + ['Reprobado'] * alumnos})
```

# Separación de clases

```python
aprob = df[df['Clase'] == 'Aprobado']
reprob = df[df['Clase'] == 'Reprobado']
```

# Visualización gráfica del conjunto de datos

```python
plt.figure(figsize = (7, 5))
plt.scatter(aprob['Horas_estudio'], aprob['Asistencia'], label = 'Aprobado', marker = 'o')
plt.scatter(reprob['Horas_estudio'], reprob['Asistencia'], label = 'Reprobado', marker = 'x')
plt.xlabel('Horas de estudio')
plt.ylabel('Asistencia')
plt.title('Distribución de alumnos')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# Centroides

```python
centroide_aprob = aprob[['Horas_estudio', 'Asistencia']].mean().values.reshape(2, 1)
centroide_reprob = reprob[['Horas_estudio', 'Asistencia']].mean().values.reshape(2, 1)
centroide_global = df[['Horas_estudio', 'Asistencia']].mean().values.reshape(2, 1)
```

# Desviación estándar de los datos se usará como radio del centroide

```python
std_aprob = aprob[['Horas_estudio', 'Asistencia']].std().mean()
std_reprob = reprob[['Horas_estudio', 'Asistencia']].std().mean()
std_global = df[['Horas_estudio', 'Asistencia']].std().mean()
```

# Distribcón de centroides

```python
plt.figure(figsize=(7, 5))
plt.scatter(aprob['Horas_estudio'], aprob['Asistencia'], label = 'Aprobado', marker = 'o')
plt.scatter(reprob['Horas_estudio'], reprob['Asistencia'], label = 'Reprobado', marker = 'x')

plt.scatter(centroide_aprob[0], centroide_aprob[1], color='blue', s = 150, edgecolors = 'black',
            label = 'Centroide Aprobado')
plt.scatter(centroide_reprob[0], centroide_reprob[1], color = 'red', s = 150, edgecolors = 'black',
            label = 'Centroide Reprobado')
plt.scatter(centroide_global[0], centroide_global[1], color = 'orange', s = 150, edgecolors = 'black',
            label = 'Centroide Global')

circle_aprob = Circle((centroide_aprob[0][0], centroide_aprob[1][0]), radius = std_aprob, fill = False,
                   edgecolor = 'blue', linestyle = '--')
circle_reprob = Circle((centroide_reprob[0][0], centroide_reprob[1][0]), radius = std_reprob, fill = False,
                    edgecolor = 'red', linestyle = '--')
circle_global = Circle((centroide_global[0][0], centroide_global[1][0]), radius = std_global, fill = False,
                       edgecolor = 'orange', linestyle = '--')

plt.gca().add_patch(circle_aprob)
plt.gca().add_patch(circle_reprob)
plt.gca().add_patch(circle_global)

plt.xlabel('Horas de estudio')
plt.ylabel('Asistencia')
plt.title('Distribución con centroides')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# Dirección de w

```python
w = inv(s_w) @ (centroide_aprob - centroide_reprob)
w_norm = w / np.linalg.norm(w)
```

```python
X = df[['Horas_estudio', 'Asistencia']].values
df['Proyeccion_LDA'] = X @ w
```

# Histograma

```python
plt.figure(figsize = (7, 5))
plt.hist(df[df['Clase'] == 'Aprobado']['Proyeccion_LDA'], bins = 10, alpha = 0.7, label = 'Aprobado')
plt.hist(df[df['Clase'] == 'Reprobado']['Proyeccion_LDA'], bins = 10, alpha = 0.7, label = 'Reprobado')
plt.xlabel('Proyección sobre la dirección de LDA')
plt.ylabel('Frecuencia')
plt.title('Proyección LDA: Separación de clases')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

# Proyección de los puntos sobre la dirección w

```python
X = df[['Horas_estudio', 'Asistencia']].values

proyecciones = X @ w_norm

puntos_proyectados = proyecciones @ w_norm.T

plt.figure(figsize = (8, 6))
plt.scatter(aprob['Horas_estudio'], aprob['Asistencia'], label = 'Aprobado', marker = 'o')
plt.scatter(reprob['Horas_estudio'], reprob['Asistencia'], label = 'Reprobado', marker = 'x')

for i in range(len(X)):
    original = X[i]
    proyectado = puntos_proyectados[i]
    plt.plot([original[0], proyectado[0]], [original[1], proyectado[1]],
             color='gray', linestyle='--', linewidth=0.8)

plt.scatter(puntos_proyectados[:, 0], puntos_proyectados[:, 1],
            color = 'purple', alpha = 0.6, label = 'Proyecciones')

plt.quiver(centroide_global[0], centroide_global[1], w_norm[0, 0], w_norm[1, 0],
           color = 'purple', scale = 5, width = 0.01, label = 'Dirección LDA (w)')

plt.xlabel('Horas de estudio')
plt.ylabel('Asistencia')
plt.title('Proyección de los puntos sobre la dirección LDA')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# Clasificación
# Calcular el umbral como el promedio de los centros proyectados

```python
media_aprob = df[df['Clase'] == 'Aprobado']['Proyeccion_LDA'].mean()
media_reprob = df[df['Clase'] == 'Reprobado']['Proyeccion_LDA'].mean()
umbral = (media_aprob + media_reprob) / 2
print(umbral)
```

# Predicciones

```python
df['Prediccion_LDA'] = df['Proyeccion_LDA'].apply(lambda x: 'Aprobado' if x >= umbral else 'Reprobado')
df['Resultado'] = df.apply(lambda row: 'Correcto' if row['Clase'] == row['Prediccion_LDA'] else 'Incorrecto', axis = 1)
df['Correcto'] = df['Clase'] == df['Prediccion_LDA']
print(df[['Alumno', 'Clase', 'Proyeccion_LDA', 'Prediccion_LDA', 'Correcto']])
```

# Función para clasificar un nuevo alumno

```python
def clasificar_nuevo_alumno(w_norm, umbral):
    try:
        horas = float(input("Ingresa las horas de estudio del alumno: "))
        asistencia = float(input("Ingresa el porcentaje de asistencia del alumno: "))
        
        nuevo = np.array([[horas, asistencia]])
        proyeccion = nuevo @ w
        
        prediccion = 'Aprobado' if proyeccion >= umbral else 'Reprobado'
        
        print(f"\nProyección LDA del nuevo alumno: {proyeccion[0][0]:.4f}")
        print(f"Clasificación: {prediccion}")
        
    except Exception as e:
        print("Error:", e)

clasificar_nuevo_alumno(w_norm, umbral)
```
