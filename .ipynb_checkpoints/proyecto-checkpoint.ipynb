```python
import pandas as pd
import matplotlib.pyplot as plt
import folium
from sklearn.neighbors import KernelDensity
from sklearn.neighbors import NearestNeighbors
import numpy as np
```

```python
df = pd.read_csv('denue_09_csv\conjunto_de_datos\denue_inegi_09_.csv', sep=',', encoding='latin1')

df.head()
df.info()
```

```python
df.shape
```

```python
# Estandarizar texto
df['nombre_act'] = df['nombre_act'].str.upper().str.strip()
df['nom_vial'] = df['nom_vial'].str.upper().str.strip()

# Eliminar nulos importantes
df = df.dropna(subset=['latitud', 'longitud'])

# Duplicados
df = df.drop_duplicates()
df.shape
```

```python
df['nombre_act'].unique()
```

```python
df['nombre_act'].value_counts()
```

```python
df['nombre_act'].value_counts().head(10).plot(kind='barh')                
plt.title('Distribución de actividades económicas (Top 10)')
plt.show()
```

```python
df['per_ocu'].value_counts()
```

```python
df['per_ocu'].value_counts().plot(kind='bar')
plt.xticks(rotation=45)
plt.title("Tamaño de los negocios")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['longitud'], df['latitud'], alpha=0.4)
plt.title("Distribución geográfica de negocios")
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.show()
```

```python
mapa = folium.Map(location=[df['latitud'].mean(), df['longitud'].mean()], zoom_start=11)

for _, row in df.sample(min(1000, len(df))).iterrows():
    folium.CircleMarker(
        location=[row['latitud'], row['longitud']],
        radius=2
    ).add_to(mapa)

mapa
```

```python
coords = df[['latitud', 'longitud']].values
nn = NearestNeighbors(radius=0.01)
nn.fit(coords)

distancias, indices = nn.radius_neighbors(coords)

df['competencia'] = [len(i) for i in indices]
```

```python
df['competencia'].hist()
plt.title("Distribución de competencia")
plt.show()
```

```python
df['lat_bin'] = pd.cut(df['latitud'], bins=20)
df['lon_bin'] = pd.cut(df['longitud'], bins=20)

df['zona'] = df['lat_bin'].astype(str) + "_" + df['lon_bin'].astype(str)
```

```python
zonas = df.groupby('zona').agg({
    'id': 'count',
    'competencia': 'mean'
}).rename(columns={'id': 'total_negocios'})
```

```python
dominante = df.groupby('zona')['nombre_act'] \
    .agg(lambda x: x.value_counts().idxmax())

zonas['actividad_dominante'] = dominante
```
