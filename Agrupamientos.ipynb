Para manejar los datos negativos con los que cuenta la variable longitud de la aleta, se realiza la imputación de sus valores usando la media de esta variable. Después se visualiza la columna coompleta y ya no cuenta con valores atípicos ni neativos.

# Agrupamiento
El agrupamiento (clustering) es una técniza de aprendizaje automático no supervisado donde se hace la agrupación de datos similares sin la necesidad de utilizar datos etiquetados. Hay dos tipos de agrupamiento principales:   
Rigida: Sucede cuando un punto de datos no puede pertenecer a varios clústeres, lo que hace que la agrupación sea clara y fácil de interpretar.  
Suave: La agrupación flexible permite que un punto de datos pertenezca a múltiples grupos con diferentes probabilidades.   
    
También hay métodos de agrupamiento de acuerdo con la manera en que se agrupan.   
Agrupamiento basado en centroides: k-means, k-medoides.  
Agrupamiento basado en densidad: SBSCAN, OPTICA.  
Agrupamiento basado en conectividad: Aglomerativo, divisivo.   
Agrupamiento basado en la distribución: modelo de mezcla gaussiana (GMM).   
Agrupamiento difuso: Fuzzy C-means.    


# Librerías
Se cargan las librerías necesarias para agrupamiento mediante K-means, DBSCAN, Gaussian Mixture (GMM), además, se utiliza PCA para redudir dimensionalidad, estandarización de datos y la métrica de Silohuette

```python
from sklearn.cluster import KMeans, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

# Escalamiento del conjunto de datos
Primero se estandarizan los datos para que las diferentes escalas de los datos no afecte a los algoritmos de agrupación.

```python
X = df_clean[numeric_cols]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

# Método del codo y Silohuette
Se aplica el método del codo y de Silohuette para encontrar el valor óptimo de K (cantidad de clústers). Para esto, se declaran dos litas vacíass, una para guardar los calores de interca y la otra para los índices de Silohuette, además se declara probar un rango para k de 2 a 10. Se realiza un ciclo for para probar cada k en el rango que se asigno, para cada ciclo se realiza la agrupación con k-means usando la cantidad k de clústeres, se entrena con los datos escalados y se guardan los datos para las listas iniciales.   
Se observa que el método de codo indica que un buen valor para k es 3, mientras que Silohuette indica que 2.

```python
inertias = []
silhouettes = []
K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(n_clusters = k, random_state = 7)
    labels = kmeans.fit_predict(X_scaled)
    inertias.append(kmeans.inertia_)
    silhouettes.append(silhouette_score(X_scaled, labels))

plt.figure(figsize = (10, 4))
plt.subplot(1, 2, 1)
plt.plot(K_range, inertias, 'bo-')
plt.xlabel("Número de clústeres (k)")
plt.ylabel("Inercia")
plt.title("Método del Codo")

plt.subplot(1, 2, 2)
plt.plot(K_range, silhouettes, 'go-')
plt.xlabel("Número de clústeres (k)")
plt.ylabel("Índice de Silhouette")
plt.title("Silhouette por k")
plt.tight_layout()
plt.show()
```

# Reducción a 2 dimensiones usando PCA para visualización
Se realiza la reducción de dimensionalidad a únicamente dos dimensiones para que se puedan visualizar los datos, de manera que se realiza mediante el análisis de componentes principales

```python
pca = PCA(n_components = 2)
X_pca = pca.fit_transform(X_scaled)
```

# K-means
K-means es un algoritmo de agrupamiento iterativo basado en centroides que divide un conjunto de datos en grupos similares en función de la distancia entre sus centroides. El centroide, o centro del clúster, es la media o la mediana de todos los puntos dentro del clúster, según las características de los datos.   
Su funcionamiento es:  
1. Se asigna un valor cualquiera de k
2. Se eligen k centroides de manera aleatoria
3. Se asigna cada punto al centroide más cercano para obtener clústeres iniciales.
4. Para cada clúster, se calcula un nuevo centroide tomando la posición media de todos los puntos del clúster.
5. Se asigna cada punto al centroide nuevo más cercano
6. Se repite para volver a calcular los centroides hasta que los puntos ya no cambien de clúster.    

En este caso, se utilizan 3 clústeres

```python
kmeans = KMeans(n_clusters = 3, random_state = 7)
labels_kmeans = kmeans.fit_predict(X_scaled)
```

# Gaussian Mixture Model
El modelo de mezcla gaussiana (GMM) es un modelo probabilístico que representa los datos como una combinación de varias distribuciones gaussianas, cada una con su propia media y varianza, ponderadas por un coeficiente de mezcla.En lugar de asumir que todos los datos provienen de una única distribución normal, un GMM asume que existen múltiples distribuciones normales, cada una de las cuales representa un grupo diferente en el conjunto de datos , y cada una con su propia media y varianza.    
De esta manera se reliza la mezcla de media, covarianza y peso de la mezcla.  
Se entrena el modelo GNN con 3 clústers

```python
gmm = GaussianMixture(n_components = 3, random_state = 7)
labels_gmm = gmm.fit_predict(X_scaled)
```

# DBSCAN
El Agrupamiento Espacial Basado en la Densidad de Aplicaciones con Ruido, agrupa puntos que están muy juntos en el espacio de datos. DBSCAN no requiere que especifiques el número de agrupamientos de antemano, lo que lo hace especialmente útil para el análisis exploratorio de datos.   

El algoritmo funciona definiendo los clústeres como regiones densas separadas por regiones de menor densidad, lo cual permite descubrir clústeres de forma arbitraria e identificar valores atípicos como ruido. Utiliza dos parámetros, epsilon, que es la distancia máxima entre dos puntos para que se consideren vecinos y los puntos mínimos que es el número mínimo de puntos necesarios para formar una región densa.

```python
dbscan = DBSCAN(eps = 0.6, min_samples = 5)
labels_dbscan = dbscan.fit_predict(X_scaled)
```

# Clasificación
La clasificación en el aprendizaje automático es un proceso de modelado predictivo mediante el cual los modelos de aprendizaje automático utilizan algoritmos de clasificación para predecir la etiqueta correcta para los datos de entrada.

# Librerías
Se importan las librerías necesarias para aplicar los algoritmos de clasificación para regresión logística, árbol de desición, k vecinos más cercanos, máquina de vectores de soporte, y bosque aleatoro. Además, se importa la función para separa el conjunto de datos en entrenamiento y prueba, las métricas de exactitud, puntuación f1, matriz de confusión, y se importa la función Lebel Encoder para transformar variables categóricas a númericas.

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn import svm
from sklearn.ensemble import RandomForestClassifier
```

# División en conjunto de entrenamiento y prueba (Con etiquetas generadas por K-Means)
Se divide el conjunto de datos en 70% entrenamiento y 30% prueba, considerando los datos escalados y las etiquetas que fueron creadas por K-means

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, labels_kmeans, test_size = 0.3, random_state = 7)
```

# Modelos de clasificación
Se utilizan los modelos de clasificaión de regresión logística, árbol de decisión, k-vecinos más cercanos, máquina de vectores de soporte, y bosque aleatorio.

```python
modelos = {"Regresión Logística": LogisticRegression(),
           "Árbol de Decisión": DecisionTreeClassifier(),
           "K-Vecinos Más Cercanos": KNeighborsClassifier(),
           "Máquina de vectores de soporte": svm.SVC(kernel='linear', C=1.0),
           "Bosque aleatorio": RandomForestClassifier(n_estimators=100)}
```

# Resultados
Se inicializa una lista vacía para guardar los resultados de cada métrica para cada modelo de clasificación, de manera que dentro del ciclo se entrena cada modelo y se calculan estas métricas. Se observa que los 5 modelos de clasificación tienen exactamente el mismo rendimiento.

```python
resultados = []

for nombre, modelo in modelos.items():
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average = 'weighted')
    resultados.append((nombre, acc, f1))

df_resultados = pd.DataFrame(resultados, columns = ["Modelo", "Accuracy", "F1-Score"])
print("Resultados de clasificación supervisada usando etiquetas de K-Means:")
print(df_resultados)
```

# Función para entrenar y evaluar clasificadores con cualquier conjunto de etiquetas
Se realiza una función para la evaluación de cada modelo de clasificación de acuerdo con las etiquetas generadas por cada algoritmo de agrupacion. De manera que se calculan las métricas de exactitud, puntuación f1, y las matrices de confusión a manera gráfica.

```python
def evaluar_modelos_con_etiquetas_y_X(X, etiquetas, nombre_metodo):
    etiquetas = LabelEncoder().fit_transform(etiquetas)
    X_train, X_test, y_train, y_test = train_test_split(X, etiquetas, test_size = 0.3, random_state = 7)

    modelos = {"Regresión Logística": LogisticRegression(),
           "Árbol de Decisión": DecisionTreeClassifier(),
           "K-Vecinos Más Cercanos": KNeighborsClassifier(),
           "Máquina de vectores de soporte": svm.SVC(kernel='linear', C=1.0),
           "Bosque aleatorio": RandomForestClassifier(n_estimators=100)}

    resultados = []
    for nombre, modelo in modelos.items():
        modelo.fit(X_train, y_train)
        y_pred = modelo.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred, average = 'weighted')
        resultados.append((nombre_metodo + " - " + nombre, acc, f1))

        cm = confusion_matrix(y_test, y_pred)
        disp = ConfusionMatrixDisplay(confusion_matrix = cm)
        disp.plot()
        plt.title(f"Matriz de confusión: {nombre_metodo} - {nombre}")
        plt.show()

    return resultados
```

# Matrices de confusión de los clasificadores con etiquetas de K-Means
Se realiza la visualización de cada matriz de confusión de cada modelo de clasificación con las etiquetas de K-means. Se observa que, los 5 clasificadores tienen el mismo rendimiento.

```python
resultados_totales = []
resultados_totales += evaluar_modelos_con_etiquetas_y_X(X_scaled, labels_kmeans, "K-Means")
```

# Matrices de confusión de los clasificadores con etiquetas de GMM
Se realiza la visualización de cada matriz de confusión de cada modelo de clasificación con las etiquetas de GMM. Se observa que, la mayoria de los clasificadores tienen el mismo rendimiento, excepto el bosque aleatorio

```python
resultados_totales += evaluar_modelos_con_etiquetas_y_X(X_scaled, labels_gmm, 'GMM')
```

# Matrices de confusión de los clasificadores con etiquetas de DBSCAN
Se realiza la visualización de cada matriz de confusión de cada modelo de clasificación con las etiquetas de K-means. Se observa que, en este caso si hay variación entre el rendimiento de los clasificadores

```python
mask_dbscan = labels_dbscan != -1
X_dbscan_filtered = X_scaled[mask_dbscan]
labels_dbscan_filtered = labels_dbscan[mask_dbscan]
resultados_totales += evaluar_modelos_con_etiquetas_y_X(X_dbscan_filtered, labels_dbscan_filtered, "DBSCAN")
```

# Resultados finales
Se imprimen los resultados de la exactitud y de la puntación f1 de cada modelo considerando las etiquetas de cada agrupamiento

```python
df_resultados_total = pd.DataFrame(resultados_totales, columns=["Modelo", "Accuracy", "F1-Score"])
print("Comparación de clasificadores entrenados con etiquetas de clustering:")
print(df_resultados_total)
```
